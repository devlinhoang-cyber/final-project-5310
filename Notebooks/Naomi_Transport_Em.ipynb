```python
# Libraries
import pandas as pd
```

```python
# Load data
emissTransport = pd.read_csv('per-capita-co2-transport.csv')
emissTransport.head()
```

```python
# Check NAs in each column
missing_values = emissTransport.isna().sum()

print(missing_values)
```

```python
emissTransport.describe()
```

```python
# Count how many times each year appears in the dataset
year_counts = emissTransport['Year'].value_counts().sort_index()
display(year_counts)
```

```python
# Group by entity and list unique years
entity_years = emissTransport.groupby('Entity')['Year'].apply(list)

# Display all
with pd.option_context('display.max_rows', None, 'display.max_colwidth', None):
    display(entity_years)
```

```python
# Define years as years without 203 year_count
target_years = set(range(1990, 2002))

# Group by Entity and check which years are missing for each
missing_years_by_entity = {}
all_entities = emissTransport['Entity'].unique()

for entity in all_entities:
    entity_years = set(emissTransport[emissTransport['Entity'] == entity]['Year'])
    missing = target_years - entity_years
    if missing:
        missing_years_by_entity[entity] = sorted(list(missing))

# Display entities missing data between 1990 and 2001
if missing_years_by_entity:
    display(pd.DataFrame(list(missing_years_by_entity.items()), columns=['Entity', 'Missing Years (1990-2001)']))
```

```python
# Truncate dataset to only include years 2010 to 2023
emissTransport = emissTransport[(emissTransport['Year'] >= 2010) & (emissTransport['Year'] <= 2023)]
```

```python
# Choose target countries
target_entities = [
    'Norway', 'China', 'United States', 'Germany',
    'South Africa', 'United Kingdom', 'Brazil', 'Thailand',
    'Japan', 'India'
]

# Filter the dataset to keep only these entities
emissTransport = emissTransport[emissTransport['Entity'].isin(target_entities)]
```

```python
emissTransport.describe()
```

```python
emissTransport = emissTransport.rename(columns={
    'Per capita carbon dioxide emissions from transport': 'Emissions from transport'
})

display(emissTransport.head())
```

```python
# Export the cleaned DataFrame to a CSV file
emissTransport.to_csv('Emission_transport_cleaned.csv', index=False)
```

### Dataset Overview (Years 2010-2023 - 10 Target Entities)

* Number of Rows: 140 rows
* What one row represents: Each row represents the per capita carbon dioxide emissions from transport in kilograms for one of the 10 selected countries in a given year between 2010 and 2023.

#### Selected Entities:
* Brazil, China, South Africa, Germany, India, Japan, Norway, Thailand, United Kingdom, and United States.

#### Columns:
1. Entity(text): The name of the country.
2. Code(text): The 3-letter country code.
3. Year(integer): The year of the observation (ranging from 2010 to 2023 inclusive).
4. Per capita carbon dioxide emissions from transport(float): The measured metric of transport related CO₂ emissions per person in kilograms.
